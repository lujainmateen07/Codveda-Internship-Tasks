```python
import pandas as pd

# Load the iris dataset
df = pd.read_csv('iris.csv')

# Show the first few rows
df.head()
```

```python
# Identify missing values and duplicates
print("Missing values:", df.isnull().sum().sum())
print("Duplicates:", df.duplicated().sum())

# Clean the data
df_cleaned = df.dropna().drop_duplicates()

# Save the cleaned file for your internship submission
df_cleaned.to_csv('Level1_Task1_Cleaned.csv', index=False)
print("Cleaning complete! Saved to 'Level1_Task1_Cleaned.csv'")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Create a simple chart for your video demonstration
plt.figure(figsize=(6,4))
sns.countplot(x='species', data=df_cleaned)
plt.title('Count of Each Species')
plt.show()
```
